# Problem Statement
An email service provider receives thousands of emails every day. Many of these emails are spam and should be filtered automatically before reaching users. Develop an NLP-based text classification model to classify SMS messages as Spam or Ham (Not Spam) using different machine learning algorithms. Compare the performance of the models using appropriate evaluation metrics

Dataset: SMS Spam Collection Dataset

https://www.kaggle.com/datasets/uciml/sms-spam-collection-dataset

```python
#Data preprocessing
#BoW, TF-IDF, Word2Vec, GloVe, FastText
#Naive Bayes, SVM, Decision Tree
#Accuracy, Precision, Recall, F1-score, Confusion Matrix
```

Import Libraries

```python
import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Download NLTK Resources
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
df=pd.read_csv("/content/spam.csv", encoding='latin-1')
```

```python
df = df[['v1', 'v2']]
df.columns = ['label', 'message']
```

```python
df['label'] = df['label'].map({'ham': 0, 'spam': 1})
```

```python
print("dataset shape:", df.shape)
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    text=text.lower()
    text=re.sub('[^a-zA-Z]',' ',text)
    words=text.split()
    words=[
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]
    return ' '.join(words)
df['clean_message'] = df['message'].apply(clean_text)
```

```python
df['clean_message']
```

Train Test Split

```python
X = df['clean_message']
y = df['label']
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42)
```

FEATURE EXTRACTION - BAG OF WORDS

```python
print("-------------Bag of Words-------------")
bow = CountVectorizer()
X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)

#Naive Bayes
nb = MultinomialNB()
nb.fit(X_train_bow, y_train)
pred_nb = nb.predict(X_test_bow)


print("\nNaive Bayes Results")
print("Accuracy :", accuracy_score(y_test, pred_nb))
print("Precision:", precision_score(y_test, pred_nb))
print("Recall   :", recall_score(y_test, pred_nb))
print("F1 Score :", f1_score(y_test, pred_nb))
print(confusion_matrix(y_test, pred_nb))
print(classification_report(y_test, pred_nb))
```

FEATURE EXTRACTION - TF = IDF


```python
print("-------------TFIDF-------------")
tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

#Support vector machine
svm = SVC()
svm.fit(X_train_tfidf, y_train)
pred_svm = svm.predict(X_test_tfidf)

print("\nSVM Results")
print("Accuracy :", accuracy_score(y_test, pred_svm))
print("Precision:", precision_score(y_test, pred_svm))
print("Recall   :", recall_score(y_test, pred_svm))
print("F1 Score :", f1_score(y_test, pred_svm))
print(confusion_matrix(y_test, pred_svm))
print(classification_report(y_test, pred_svm))

#Decision tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train_tfidf, y_train)
pred_dt = dt.predict(X_test_tfidf)

print("\nDecision Tree Results")
print("Accuracy :", accuracy_score(y_test, pred_dt))
print("Precision:", precision_score(y_test, pred_dt))
print("Recall   :", recall_score(y_test, pred_dt))
print("F1 Score :", f1_score(y_test, pred_dt))
print(confusion_matrix(y_test, pred_dt))
print(classification_report(y_test, pred_dt))
```

MODEL COMPARISION

```python
results = pd.DataFrame({
    "Model":["Naive Bayes","Support Vector Machine","Decision Tree"],
    "Accuracy":[
        accuracy_score(y_test,pred_nb),
        accuracy_score(y_test,pred_svm),
        accuracy_score(y_test,pred_dt)
    ],
    "Precision":[
        precision_score(y_test,pred_nb),
        precision_score(y_test,pred_svm),
        precision_score(y_test,pred_dt)
    ],
    "Recall":[
        recall_score(y_test,pred_nb),
        recall_score(y_test,pred_svm),
        recall_score(y_test,pred_dt)
    ],
    "F1 Score":[
        f1_score(y_test,pred_nb),
        f1_score(y_test,pred_svm),
        f1_score(y_test,pred_dt)
    ]
})

print("\n========== MODEL COMPARISON ==========")
print(results)
```

PREDICT NEW MESSAGE

```python
sample = [
    "Congratulations! You have won a free iPhone. Click here to claim your prize."
]

sample = [clean_text(text) for text in sample]

sample_vector = tfidf.transform(sample)

prediction = svm.predict(sample_vector)

print("\n========== SAMPLE PREDICTION ==========")

if prediction[0] == 1:
    print("Message:", sample[0])
    print("Prediction: SPAM")
else:

    print("Prediction: HAM")
```
